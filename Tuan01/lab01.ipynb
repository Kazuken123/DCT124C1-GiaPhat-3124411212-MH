## Bài 6. Phép toán và nhân ma trận


```python
# //Bai 6
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:\n", a)
print("b:\n", b)
print("\na + b:\n", a + b)
print("\na / b:\n", a / b)
print("\na^2:\n", a ** 2)
print("\nMatrix multiplication (a @ b.T):\n", a @ b.T)
print("Shape a @ b.T:", (a @ b.T).shape)
```

**Yêu cầu 2:** Cho biết shape của kết quả a @ b.T.

## Bài 7. Các hàm thống kê trên Tensor


```python
# //Bai 7
import torch

a = torch.randn(3, 4)
print("a:\n", a)

print("\n--- dim=0 (theo từng cột) ---")
print("Mean:", torch.mean(a, dim=0))
print("Standard deviation:", torch.std(a, dim=0))
print("Cumulative sum:\n", torch.cumsum(a, dim=0))

print("\n--- dim=1 (theo từng hàng) ---")
print("Mean:", torch.mean(a, dim=1))
print("Standard deviation:", torch.std(a, dim=1))
print("Cumulative sum:\n", torch.cumsum(a, dim=1))
```

### Yêu cầu Bài 7
**Yêu cầu:** Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.  
**Trả lời:**  
- `dim=0`: Tính toán gom theo từng cột (kết quả có 4 giá trị tương ứng với 4 cột).  
- `dim=1`: Tính toán gom theo từng hàng (kết quả có 3 giá trị tương ứng với 3 hàng).


## Bài 8. Autograd cơ bản


```python
# //Bai 8
import torch

x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()
print("x2.grad =", x2.grad)
```

### Yêu cầu Bài 8
**Yêu cầu 1:** Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với x.grad.  
**Trả lời:**  
- Đạo hàm: $y' = 2x$.  
- Tại $x = 3.6$: $y'(3.6) = 2 \times 3.6 = 7.2$.  
- Kết quả tính thủ công bằng chính xác với `x.grad = 7.2`.

**Yêu cầu 2:** Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.  
**Trả lời:**  
- Đạo hàm: $y' = 6x + 2$.  
- Tại $x = 3.6$: $y'(3.6) = 6 \times 3.6 + 2 = 23.6$.  
- Kết quả `x2.grad = 23.6` hoàn toàn chính xác.


## Bài 9. Gradient Descent tìm hệ số đa thức


```python
# //Bai 9
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:\n", w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:\n", w)
```

### Yêu cầu Bài 9
**Yêu cầu 1:** So sánh ba hệ số học được với [1, 2, 3].  
**Trả lời:** Ba hệ số tối ưu học được trong `w` tiệm cận rất sát với các hệ số gốc [1, 2, 3].

**Yêu cầu 2:** Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().  
**Trả lời:**  
- `optimizer.zero_grad()`: Đặt gradient tích lũy về 0 trước khi tính lượt mới (tránh việc gradient bị cộng dồn).  
- `mse.backward()`: Lan truyền ngược để tính đạo hàm riêng (gradient) của hàm mất mát mse theo trọng số w.  
- `optimizer.step()`: Cập nhật giá trị trọng số w theo gradient vừa tính và learning rate.


## Bài 10. Giải hệ phương trình bằng Autograd


```python
# //Bai 10
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print(f"A = {A.item():.4f}")
print(f"B = {B.item():.4f}")
print(f"C = {C.item():.4f}")
print(f"D = {D.item():.4f}")

print("\nKiểm tra lại hệ phương trình:")
print("A + B =", (A + B).item())
print("C - D =", (C - D).item())
print("A + C =", (A + C).item())
print("B - D =", (B - D).item())
```

### Yêu cầu Bài 10
**Yêu cầu 1:** Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.  
**Trả lời:** Nghiệm thu được thỏa mãn cả 4 phương trình với sai số xấp xỉ bằng 0.

**Yêu cầu 2:** Nhận xét vì sao bài toán có thể có nhiều nghiệm.  
**Trả lời:** Hệ gồm 4 ẩn nhưng chỉ có 3 phương trình độc lập tuyến tính, do $(A+B) - (A+C) + (C-D) = B-D$ ($9 - 8 + 1 = 2$). Ma trận hệ số có rank = 3 < 4 (số ẩn) nên hệ phương trình có vô số nghiệm.


## Bài 11. Bài tập áp dụng


```python
# //Bai 11
import numpy as np
import torch

# 1. Tạo Tensor X 24 phần tử và reshape
X = torch.arange(24)
print("X ban đầu:\n", X)
print("X reshape 4 x 6:\n", X.reshape(4, 6))
print("X reshape 2 x 12:\n", X.reshape(2, 12))
print("X reshape 2 x 3 x 4:\n", X.reshape(2, 3, 4))

# 2. Hai Tensor A, B kích thước 3 x 4
A = torch.randn(3, 4)
B = torch.rand(3, 4) + 0.1
print("\nA:\n", A)
print("B:\n", B)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung bình theo cột của A:", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột của A:", torch.std(A, dim=0))

# 3. y = 2x^2 + 5x + 3 tại x = 2
x3 = torch.tensor(2.0, requires_grad=True)
y3 = 2 * x3**2 + 5 * x3 + 3
y3.backward()
print("\nAutograd dy/dx tại x=2:", x3.grad.item())

# 4. Huấn luyện y = 2x^2 - 3x + 5
poly11 = np.poly1d([2, -3, 5])
N11 = 50
X11 = np.random.randn(N11, 1) * 5
Y11 = poly11(X11)
XX11 = np.hstack([X11 * X11, X11, np.ones_like(X11)])

w11 = torch.randn(3, 1, requires_grad=True)
x11 = torch.tensor(XX11, dtype=torch.float32)
y11 = torch.tensor(Y11, dtype=torch.float32)
opt11 = torch.optim.NAdam([w11], lr=0.01)

for epoch in range(2000):
    opt11.zero_grad()
    y_pred11 = x11 @ w11
    loss = torch.mean(torch.square(y11 - y_pred11))
    loss.backward()
    opt11.step()

print("\nHệ số tìm được:\n", w11)
```

### Yêu cầu Bài 11
**Yêu cầu 1:** Dùng autograd tính dy/dx của y = 2x² + 5x + 3 tại x = 2 và kiểm tra bằng phép tính thủ công.  
**Trả lời:**  
- Đạo hàm: $dy/dx = 4x + 5$.  
- Tại $x = 2$: $dy/dx = 4 \times 2 + 5 = 13$.  
- Khớp hoàn toàn với kết quả autograd tính được là 13.

**Yêu cầu 2:** Cho biết các hệ số tìm được khi đổi đa thức thành y = 2x² - 3x + 5.  
**Trả lời:** Các hệ số học được tiệm cận xấp xỉ [2, -3, 5].
